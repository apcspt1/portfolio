### Tier 1 of Learning: Distraction Management and Time Spacing
<span style="color: #ff8513; font-size: 20px; font-family: Quicksand;">*Distraction Management*:</span>
- *always* ensure your study environment is neat and tidy, otherwise clean up all messes, to increase focus on studies

- *silence* notifications and put away any distractions like devices or fidgets in another room

- *work* in a workspace that suits you best, whether that's your bedroom, a library, coffee shop, or in a calm outdoor space

<br>

<span style="color: #ff8513; font-size: 20px; font-family: Quicksand;">*Pomodoro Technique*:</span>
- *work* in dedicated 25-minute focus intervals followed by short 5-minute breaks to keep your brain refreshed and prevent burnout

- *during* each 25-minute cycle, commit exclusively to one task without switching tabs, checking messages, or breaking focus.

- *after* completing 4 consecutive Pomodoro cycles, reward yourself with a longer 15-to-30-minute break to fully recharge before starting the next block.

- *keep* a simple tally or timer log of how many cycles you complete each day to measure your focus consistency over time.

<br>

```python
# CODE_RUNNER: Tier 1 of Learning

# USER: Input the hours you want to focus studying below
planned_study_minutes = 0;  # e.g., 2 hours

pomodoro_session_length = 25
break_length = 5

total_sessions = planned_study_minutes // (pomodoro_session_length + break_length)
total_focus_time = total_sessions * pomodoro_session_length
total_break_time = total_sessions * break_length

# 3. Log results to console
print("=== TIER 1: FOCUS & ENVIRONMENT PLANNER ===")
print(f"Planned Time: {planned_study_minutes} minutes")
print("-" * 39)
print(f"Full Pomodoro Sessions: {total_sessions}")
print(f"Pure Focus Time: {total_focus_time} mins")
print(f"Rest Breaks Earned: {total_break_time} mins")

if total_sessions >= 4:
    print("\nSTATUS: TIER 1 MASTER! Remember to take a longer 15-minute break after 4 sessions.")
else:
    print("\nSTATUS: Tier 1 Initiate. Complete your sessions to unlock Tier 2!") 
```
