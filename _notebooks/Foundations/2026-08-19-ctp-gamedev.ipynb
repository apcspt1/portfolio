**Project Purpose:** 
MindForge is an interactive tool designed to help you analyze, track, and optimize your study habits. Instead of focusing on what you study, MindForge focuses on how you study. By turning cognitive science principles into executable code, this project measures your active learning habits, focus levels, and mastery progression.


### How the System Works
As you use MindForge, you progress through three core levels of metacognition:

*   **Tier 1 (Foundational Habits):** Establishing baseline focus techniques like Pomodoro cycles and distraction management.
*   **Tier 2 (Deep Processing):** Applying active recall, error logging, and the Feynman technique to retain complex information.
*   **Tier 3 (Mastery & Synthesis):** Interleaved subjects, self-testing, and peer Teaching.

function runProfileCode() {
  let consoleDiv = document.getElementById("runnerConsole");
  consoleDiv.innerText = ""; 
  
  function DISPLAY(text) {
    consoleDiv.innerText += text + "\n";
  }

  try {
    let code = document.getElementById("runnerCodeText").value;
    eval(code);
  } catch (err) {
    consoleDiv.innerText = "Execution Error: " + err.message;
  }
}
